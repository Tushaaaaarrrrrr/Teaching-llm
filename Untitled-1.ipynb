```python
# Python built-in data types: examples
integer_value = 10
float_value = 3.14
complex_value = 2 + 3j
string_value = "hello"
list_value = [1, 2, 3]
tuple_value = (1, 2, 3)
range_value = range(5)
dict_value = {"name": "Asha", "age": 20}
set_value = {1, 2, 3}
frozenset_value = frozenset({1, 2, 3})
bool_value = True
bytes_value = b"abc"
bytearray_value = bytearray(b"abc")
none_value = None

examples = [
    ("int", integer_value),
    ("float", float_value),
    ("complex", complex_value),
    ("str", string_value),
    ("list", list_value),
    ("tuple", tuple_value),
    ("range", range_value),
    ("dict", dict_value),
    ("set", set_value),
    ("frozenset", frozenset_value),
    ("bool", bool_value),
    ("bytes", bytes_value),
    ("bytearray", bytearray_value),
    ("NoneType", none_value),
]

for name, value in examples:
    print(f"{name:10} -> {value!r} | type: {type(value).__name__}")
```
